```python
import pandas as pd 
from sqlalchemy import create_engine, text

engine = create_engine(
    "mysql+pymysql://root:Rajuujar91929394@localhost/sql_performance_benchmarking"
)
```

```python
customers = pd.read_csv(r"D:\python project files\Sql_Query_Optimization_Performance_Benchmarking\data\raw\customers.csv")
customers.to_sql("customers",engine,if_exists="append",index=False)
print("customers loaded")
```

```python
csv_file = (r"D:\python project files\Sql_Query_Optimization_Performance_Benchmarking\data\raw\order_items.csv")
chunksize = 100000

# Track how many rows already loaded
with engine.connect() as conn:
    result = conn.execute(text("SELECT COUNT(*) FROM order_items"))
    already_loaded = result.scalar()

print(f"Already loaded {already_loaded} rows")

# Skip rows already loaded
skiprows = range(1, already_loaded+1) if already_loaded > 0 else None

for chunk in pd.read_csv(csv_file, chunksize=chunksize, skiprows=skiprows):
    chunk.to_sql("order_items", con=engine, if_exists="append", index=False, method="multi")
    already_loaded += len(chunk)
    print(f"{already_loaded} rows loaded so far")
```
