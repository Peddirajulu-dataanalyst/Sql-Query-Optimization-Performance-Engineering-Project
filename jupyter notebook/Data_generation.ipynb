```python
# logic for generate customers.csv
Rows: 100,000

customer_id       → Unique
customer_name     → Faker Name
city              → Faker City
state             → Faker State
registration_date → Last 10 years
customer_segment  → Consumer / Corporate / Small Business
```

```python
# script for generate customers.csv

import pandas as pd
from faker import Faker
import random

fake = Faker()

num_customers = 100000

customers = []

segments = [
    "Consumer",
    "Corporate",
    "Small Business"
]

for customer_id in range(1, num_customers + 1):

    customers.append({

        "customer_id": customer_id,

        "customer_name": fake.name(),

        "city": fake.city(),

        "state": fake.state(),

        "registration_date": fake.date_between(
            start_date="-10y",
            end_date="today"
        ),

        "customer_segment": random.choice(
            segments
        )
    })

customers_df = pd.DataFrame(customers)

customers_df.to_csv(
    r"D:\python project files\Sql_Query_Optimization_Performance_Benchmarking\data\raw\customers.csv",
    index=False
)

print(customers_df.head())
print("customers.csv created")
```

```python
#script for products.csv

import pandas as pd
import random

num_products = 20000

categories = [
    "Electronics",
    "Furniture",
    "Clothing",
    "Sports",
    "Beauty",
    "Office Supplies",
    "Home & Kitchen",
    "Toys"
]

products = []

for product_id in range(1, num_products + 1):

    products.append({

        "product_id": product_id,

        "product_name": f"Product_{product_id}",

        "category": random.choice(
            categories
        ),

        "price": round(
            random.uniform(10, 5000),
            2
        )
    })

products_df = pd.DataFrame(products)

products_df.to_csv(
    r"D:\python project files\Sql_Query_Optimization_Performance_Benchmarking\data\raw\products.csv",
    index=False
)

print(products_df.head())
print("products.csv created")
```

```python
#logic for geerating orders.csv

Rows: 500,000

Every order belongs to a customer.
```

```python
# script for generating orders.csv

import pandas as pd
import random
from faker import Faker

fake = Faker()

num_orders = 500000

payment_methods = [
    "Credit Card",
    "Debit Card",
    "UPI",
    "PayPal",
    "Bank Transfer"
]

orders = []

for order_id in range(1, num_orders + 1):

    orders.append({

        "order_id": order_id,

        "customer_id": random.randint(
            1,
            100000
        ),

        "order_date": fake.date_between(
            start_date="-3y",
            end_date="today"
        ),

        "order_amount": round(
            random.uniform(50, 5000),
            2
        ),

        "payment_method": random.choice(
            payment_methods
        )
    })

orders_df = pd.DataFrame(orders)

orders_df.to_csv(
    r"D:\python project files\Sql_Query_Optimization_Performance_Benchmarking\data\raw\orders.csv",
    index=False
)

print(orders_df.head())
print("orders.csv created")
```

```python
#logc for generating order_item.csv
Rows: 1,500,000

Each row represents one product
inside an order.
```

```python
# script for generating order_item.csv

import pandas as pd
import random

num_order_items = 1500000

order_items = []

for item_id in range(
    1,
    num_order_items + 1
):

    quantity = random.randint(
        1,
        5
    )

    unit_price = random.uniform(
        10,
        5000
    )

    sales = round(
        quantity * unit_price,
        2
    )

    profit = round(
        sales * random.uniform(
            0.10,
            0.30
        ),
        2
    )

    order_items.append({

        "order_item_id": item_id,

        "order_id": random.randint(
            1,
            500000
        ),

        "product_id": random.randint(
            1,
            20000
        ),

        "quantity": quantity,

        "sales": sales,

        "profit": profit
    })

order_items_df = pd.DataFrame(
    order_items
)

order_items_df.to_csv(
    r"D:\python project files\Sql_Query_Optimization_Performance_Benchmarking\data\raw\order_items.csv",
    index=False
)

print(order_items_df.head())
print("order_items.csv created")
```

```python
#logic for generating web_sessions.csv

Rows: 1,000,000

Customer website activity.
```

```python
# script for generating web_sessions.csv

import pandas as pd
import random
from faker import Faker

fake = Faker()

num_sessions = 1000000

devices = [
    "Mobile",
    "Desktop",
    "Tablet"
]

sessions = []

for session_id in range(
    1,
    num_sessions + 1
):

    sessions.append({

        "session_id": session_id,

        "customer_id": random.randint(
            1,
            100000
        ),

        "session_date": fake.date_between(
            start_date="-2y",
            end_date="today"
        ),

        "device_type": random.choice(
            devices
        ),

        "session_duration": random.randint(
            30,
            1800
        ),

        "page_views": random.randint(
            1,
            25
        )
    })

sessions_df = pd.DataFrame(
    sessions
)

sessions_df.to_csv(
    r"D:\python project files\Sql_Query_Optimization_Performance_Benchmarking\data\raw\web_sessions.csv",
    index=False
)

print(sessions_df.head())
print("web_sessions.csv created")
```
